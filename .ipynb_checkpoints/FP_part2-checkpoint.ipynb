### Estimating Error on the CMB Optical Depth from Cleaning with WMAP and Planck

As a final project for the class, we will take a first pass at a problem that we need to solve for the CLASS project. CLASS aims to measure the optical depth to reionization ($\tau$) to its cosmic variance limit on 75% of the sky that CLASS measures. To do this we use low frequency data from WMAP and high frequency data from Planck to clean polarized synchrotron and dust emission, respectively, from our data at 93 GHz. We need to understand to what extent using the WMAP and Planck data limits us in achieving the goal of a cosmic-variance measurement of $\tau$. 

* Even though CLASS only measures 75% of the sky, in this project we will use all-sky maps to simplify the analysis.
* For simulations, use `camb` to generate angular power spectra given cosmological parameters and `healpy` `synfast` to generate maps from these spectra. For input cosmological parameters use those introduced in the `get_spectra` function of Lecture 5.
* Use `healpy` `anafast` to estimate auto and cross power spectra. 
* For foreground models, let's use `pysm`'s `s1` model for synchrotron and its `d1` model for dust.
* The low-$\ell$ ($\ell<32$) E-mode spectrum of the CMB polarization increases proportional to $\tau^2$. When estimating a best-fit $\tau$ you will use a likelihood built from a simulated E-mode angular power spectrum (the data) for $\ell<32$ and a model for the power spectrum generated by `camb` (Lecture 8).
* For simplicity, assume a Gaussian likelihood for the $C_\ell$ with no covariance between $C_\ell$. If there is time, we'll revisit this assumption in the light of arguments like the one in problem 3 of lecture 8.
* When fitting the model for $C_\ell$, you can hold all cosmological parameters but $\tau$ fixed to their input values (second bullet above). Because adjusting $\tau$ changes $C_\ell$, and therefore the cosmic variance, you will need to adjust the variance in your likelihood as well.
* You should be able to get away with NSIDE=32 healpix maps.

### Part 2:

When we introduce white noise and then take the auto and cross spectra, we must treat things a little differently.

You can reference the [PS and likelihood slide deck](https://www.cpt.univ-mrs.fr/~cosmo/EC2017/Presentations/Hivon-1.pdf) that Dr. Marraige provided (slide 12) to see that the mean and variance of the auto spectrum is
$$
\langle\hat C_\ell \rangle = C_\ell(\tau) + N_{\ell},\qquad
\mathrm{var}(\hat C_\ell)=\frac2{2\ell+1}(C_\ell(\tau) + N_{\ell})^2
$$

which then requires us to minimize

$$
\frac12\chi^2+\frac12\sum_i\ln(2\pi\sigma_i(\tau)^2)\equiv \boxed{\frac12\sum_\ell\frac{((C_\ell(\tau) + N_{\ell})-y_\ell)^2}{\frac{2}{2\ell+1}(C_\ell(\tau) + N_{\ell})^2}+\ln[\frac{4\pi}{2\ell+1}(C_\ell(\tau) + N_{\ell})^2]}
$$

For the cross spectrum, you can reference the same slide to see that the mean and variance would be
$$
\langle\hat C_\ell \rangle = C_\ell(\tau),\qquad
\mathrm{var}(\hat C_\ell)=\frac1{2\ell+1}[C_\ell(\tau)^2 + (C_\ell(\tau) + N_{1, \ell})*(C_\ell(\tau) + N_{2, \ell})].
$$

Where $N_{1, \ell} = N_{2, \ell} = 2N_{\ell}$, leaving the variance to be

$$
\mathrm{var}(\hat C_\ell)=\frac2{2\ell+1}(C_\ell(\tau)^2 + 2C_\ell(\tau)N_{\ell} + 2N_{\ell}^2)
$$

which then requires us to minimize

$$
\frac12\chi^2+\frac12\sum_i\ln(2\pi\sigma_i(\tau)^2)\equiv \boxed{\frac12\sum_\ell\frac{(C_\ell(\tau)-y_\ell)^2}{\frac2{2\ell+1}(C_\ell(\tau)^2 + 2C_\ell(\tau)N_{\ell} + 2N_{\ell}^2}+\ln[2\pi\frac2{2\ell+1}(C_\ell(\tau)^2 + 2C_\ell(\tau)N_{\ell} + 2N_{\ell}^2)]}
$$


```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import healpy as hp
import os
import camb
from camb import model, initialpower
from scipy.optimize import minimize
from corner import corner

pars = camb.CAMBparams()
```

```python
lmax=32
def get_Clhat(tau, lmax=lmax, raw_cl=True,
              H0=67, Ombh2=0.024, Omch2=0.12,
              As=2.1e-9, ns=0.965, ks=0.05, 
              kt=0.05, Omk=0, TCMB=2.7255, 
              mnu=0.06, r=0, YHe=0.245400,):
    '''
    Specify tau, and use other known parameters
    '''
    pars.set_cosmology(H0=H0, ombh2=Ombh2, omch2=Omch2, mnu=mnu, omk=Omk, tau=tau, TCMB=TCMB, YHe=YHe)
    pars.InitPower.set_params(As=As, ns=ns, r=r, pivot_scalar=ks, pivot_tensor=kt)
    results = camb.get_results(pars)
    powers = results.get_total_cls(lmax=lmax, CMB_unit='muK', raw_cl=True)
    TT, EE, BB, TE = powers.T
    #ls = np.arange(2, EE.shape[0]) # could add??
    return np.array([TT, EE, BB, TE])

true_tau = 0.054
ls = np.arange(2, lmax+1)
true_spectra = get_Clhat(true_tau) # ensemble average power spectrum
Z = ls*(ls+1)/(2*np.pi)
plt.semilogx(ls, true_spectra[1][ls]*Z)
plt.xlabel(r'$\ell$')
plt.ylabel(r'$\frac{\ell(\ell+1)}{2\pi}C_\ell$\,$\mathrm{[\mu K^2]}$')
plt.title("True CMB EE Spectrum")
```

```python
nside = 32
npix = hp.nside2npix(nside)
solid_angle = hp.nside2pixarea(nside)
# map sensitivity
noise = np.array([10, 20, 40, 80]) #uK arcmin
sigma_pix = noise * (np.pi / 180 / 60)/ np.sqrt(solid_angle)
print(sigma_pix)
```

```python
np.random.seed(137)

s = hp.synfast(true_spectra, nside, lmax=lmax, new=True)

n1 = np.random.randn(npix) * sigma_pix[3]
n2 = np.random.randn(npix) * sigma_pix[3]

m1 = s + n1 * np.sqrt(2)
m2 = s + n2 * np.sqrt(2)
m_auto = s + n1

almT,almE,almB = hp.map2alm(m1,lmax=lmax)
E_map = hp.alm2map(almE, nside)
hp.mollview(E_map, min=-2, max=2, unit=r'$\mathrm{\mu K}$', cmap='RdBu_r', title='CMB Realization', xsize=2000)
almT,almE,almB = hp.map2alm(m2,lmax=lmax)
E_map = hp.alm2map(almE, nside)
hp.mollview(E_map, min=-2, max=2, unit=r'$\mathrm{\mu K}$', cmap='RdBu_r', title='CMB Realization', xsize=2000)
almT,almE,almB = hp.map2alm(m12,lmax=lmax)
E_map = hp.alm2map(almE, nside)
hp.mollview(E_map, min=-2, max=2, unit=r'$\mathrm{\mu K}$', cmap='RdBu_r', title='CMB Realization', xsize=2000)
```

```python
Clhat_cross = hp.anafast(m1, m2, lmax=lmax)[1]
Clhat12 = hp.anafast(m_auto, lmax=lmax)[1]

plt.semilogx(ls, Clhat_cross[ls]*Z, label='Cross-spectra')
plt.semilogx(ls, Clhat12[ls]*Z, label='auto spectra')
plt.semilogx(ls, true_spectra[1][ls]*Z, label='True spectra')
plt.legend()
plt.xlabel(r'$\ell$')
plt.ylabel(r'$\frac{\ell(\ell+1)}{2\pi}C_\ell$,$\mathrm{[\mu K^2]}$')
plt.title('Cross-spectra and auto spectra')
```

```python
# finding best fit tau for Clhat by minimizing chi square
cosmic_var = np.sqrt(2/(2*ls + 1))*true_spectra[1][ls]
def make_chi2(Clhat):

    def chi2(tau):
        theory_EE = get_Clhat(tau)[1]
        return sum(((Clhat[2:33] - theory_EE[2:33])/cosmic_var)**2)
    
    return chi2
chi2 = make_chi2(Clhat_cross)
sol = minimize(chi2, true_tau, bounds=[(0.04, 0.2)])
print(sol.x)
```

```python
def run_simulation_grid(EE_grid, nsim, noise_level):
    tau_auto = []
    tau_cross = []

    # var_grid_auto = np.array([
    #     (2/(2*ls + 1))*(ee[ls] + Nl[noise_level]/2)**2
    #     for ee in EE_grid
    # ])

    # POSSIBLY CORRECTED
    var_grid_auto = np.array([
        (2/(2*ls + 1))*(ee[ls] + Nl[noise_level])**2
        for ee in EE_grid
    ])

    # var_grid_cross = np.array([
    #     (1/(2*ls + 1))*(2*(ee[ls]**2) + (Nl[noise_level])**2 + (2*ee[ls]*Nl[noise_level]))
    #     for ee in EE_grid
    # ])

    # POSSIBLY CORRECTED
    var_grid_cross = np.array([
        (1/(2*ls + 1))*(2*(ee[ls]**2) + 4*(Nl[noise_level])**2 + (4*ee[ls]*Nl[noise_level]))
        for ee in EE_grid
    ])

    log_var_auto = np.log(2*np.pi*var_grid_auto)
    log_var_cross = np.log(2*np.pi*var_grid_cross)
    
    for i in range(nsim):
        s = hp.synfast(true_spectra, nside, lmax=lmax, new=True)

        n1 = np.random.randn(npix) * sigma_pix[noise_level] 
        n2 = np.random.randn(npix) * sigma_pix[noise_level] 

        m1 = s + n1 * np.sqrt(2)
        m2 = s + n2 * np.sqrt(2)
        m_auto = s + n1

        Clhat_cross = hp.anafast(m1, m2, lmax=lmax)[1]
        Clhat_auto = hp.anafast(m_auto, lmax=lmax)[1]

        chi2_cross = np.sum(
            ((Clhat_cross[None, 2:33] - EE_grid[:, 2:33])**2 / var_grid_cross) + log_var_cross,
            axis=1
        )

        chi2_auto = np.sum(
            ((Clhat_auto[None, 2:33] - (Nl[noise_level]) - EE_grid[:, 2:33])**2 / var_grid_auto) + log_var_auto,
            axis=1
        )

        tau_cross.append(tau_grid[np.argmin(chi2_cross)])
        tau_auto.append(tau_grid[np.argmin(chi2_auto)])

    return np.array(tau_auto), np.array(tau_cross)
```

```python
def run_simulation(noise_level):
    tau_auto = []
    tau_cross = []
    for i in range(100):
        s = hp.synfast(true_spectra, nside, lmax=lmax, new=True)

        n1 = np.random.randn(npix) * sigma_pix[noise_level] 
        n2 = np.random.randn(npix) * sigma_pix[noise_level] 

        m1 = s + n1 * np.sqrt(2)
        m2 = s + n2 * np.sqrt(2)
        m12 = (2*s + n1 + n2) / 2

        Clhat_cross = hp.anafast(m1, m2, lmax=lmax)[1]
        Clhat12 = hp.anafast(m12, lmax=lmax)[1]

        chi2_cross = make_chi2(Clhat_cross)
        chi2_auto = make_chi2(Clhat12 - Nl[noise_level]/2) 

        sol_cross = minimize(chi2_cross, true_tau, bounds=[(0.04, 0.2)])
        tau_cross.append(sol_cross.x[0])
        sol_auto = minimize(chi2_auto, true_tau, bounds=[(0.04, 0.2)])
        tau_auto.append(sol_auto.x[0])

        if i % 10 == 0:
            print(f"Progress: {i}/100")
    tau_auto = np.array(tau_auto)
    tau_cross = np.array(tau_cross)
    return tau_auto, tau_cross
```

```python
tau_auto_10, tau_cross_10 = run_simulation_grid(EE_grid, 1000, 0)

print(f"{'Type':<28} {'Std':>12} {'Mean':>12}")
print("-" * 54)
print(f"{'Auto Spectrum':<28} {tau_auto_10.std():>12.6f} {tau_auto_10.mean():>12.6f}")
print(f"{'Cross Spectrum':<28} {tau_cross_10.std():>12.6f} {tau_cross_10.mean():>12.6f}")
```

```python
def log_posterior(tau, ell, theory_EE, noise_level):
    return log_like(tau, ell, theory_EE, noise_level) + log_prior(tau, ell)

def log_like(tau, ell, theory_EE, noise_level):
    chi2 = sum(((true_spectra[1][ell] - theory_EE[ell])/(cosmic_var[ell]+noise_level))**2)
    lndet = sum(np.log(2*np.pi*(cosmic_var[ell]+noise_level)**2))
    return -0.5*chi2 - 0.5*lndet

def log_prior(tau, ell):
    # Using a flat prior for simplicity - pdf = 1 without normalization
    # if (tau < 0.04) or (tau > 0.2):
    if tau < 0:
        return -np.inf
    else:
        return 0
    
ls = np.arange(lmax+1)
cosmic_var = np.sqrt(2/(2*ls + 1))*true_spectra[1][ls]
```

```python
from numpy.random import default_rng
seed = 137
rng = default_rng(seed)

nsteps = 2000
# chain = np.zeros((nsteps, len(sol.x)))
guess = 0.054
# chain[0] = guess #sol.x

ell = np.arange(2, lmax+1)
def run_mcmc(noise_level):
    chain = np.zeros((nsteps, len(sol.x)))
    chain[0] = guess #sol.x
    for t in range(1,nsteps):
        x_test = rng.standard_normal()*0.003 + chain[t-1]
        theory_new = get_Clhat(x_test)[1]
        theory_old = get_Clhat(chain[t-1])[1]
        A = np.exp(log_posterior(x_test,ell,theory_new, noise_level) - log_posterior(chain[t-1],ell, theory_old, noise_level))
        accept = (A > rng.random())
        if accept:
            chain[t] = x_test
        else:
            chain[t] = chain[t-1]
        if t % 100 == 0:
            print(t)
    print("done")
    return chain
```
